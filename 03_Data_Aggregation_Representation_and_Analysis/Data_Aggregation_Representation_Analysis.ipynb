# **📊 Data Aggregation & Representation**

# Load Cleaned Dataset

```python
import pandas as pd

df = pd.read_csv('/content/paper_leaks_cleaned_final.csv')

pd.set_option('display.max_rows', None)

print("Cleaned dataset loaded successfully")
print("Shape:", df.shape)

display(df.head())
```

# Create Year Column

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['year'] = df['date'].dt.year

print("Year column created successfully")
```

# 1. How many paper leaks occurred each year?

```python
# Paper Leaks by Year
df['year'] = pd.to_datetime(df['date']).dt.year

yearly_leaks = df.groupby('year').size().reset_index(name='leak_count')

display(yearly_leaks)
```

# 2. How many paper leaks occurred in each era?

```python
# Paper Leaks by Era

era_summary = df.groupby('era').agg(
    leak_count=('incident_id', 'count')
).reset_index()

display(era_summary)
```

# 3. Which areas had the highest number of paper leaks?

```python
# Paper Leaks by Area

area_summary = df.groupby('area').agg(
    leak_count=('incident_id', 'count')
).reset_index()

area_summary = area_summary.sort_values(
    'leak_count', ascending=False
)

display(area_summary)
```

# 4. What is the yearly total of affected aspirants, arrests, and convictions?

```python
# Yearly Summary of Impact and Legal Action

yearly_summary = df.groupby('year').agg(
    total_aspirants=('aspirants_affected', 'sum'),
    total_arrests=('arrests', 'sum'),
    total_convictions=('convictions', 'sum')
).reset_index()

display(yearly_summary)
```

# 5. How many cases belong to each leak status?

```python
# Cases by Leak Status

status_summary = df.groupby('leak_status').size().reset_index(
    name='case_count'
)

display(status_summary)
```

# 6. What is the overall impact of paper leaks?

```python
# Overall Impact Summary

impact_summary = pd.DataFrame({
    'Metric': [
        'Total Paper Leak Cases',
        'Total Affected Aspirants',
        'Total Arrests',
        'Total Convictions',
        'Total Linked Deaths'
    ],
    'Total': [
        len(df),
        df['aspirants_affected'].sum(),
        df['arrests'].sum(),
        df['convictions'].sum(),
        df['linked_deaths'].sum()
    ]
})

display(impact_summary)
```

# **📈 Data Analysis**

# 1. Which year had the highest number of paper leak cases?

```python
# Highest Paper Leak Year

year_analysis = df.groupby('year').size().reset_index(name='leak_count')

highest_year = year_analysis.loc[
    year_analysis['leak_count'].idxmax()
]

display(highest_year)
```

# 2. Which area had the highest number of paper leak cases?

```python
# Area with Highest Paper Leaks

area_analysis = df.groupby('area').size().reset_index(name='leak_count')

highest_area = area_analysis.loc[
    area_analysis['leak_count'].idxmax()
]

display(highest_area)
```

# 3. Which exam had the highest number of paper leak cases?

```python
# Exam with Highest Paper Leaks

exam_analysis = df.groupby('exam_name').size().reset_index(name='leak_count')

highest_exam = exam_analysis.loc[
    exam_analysis['leak_count'].idxmax()
]

display(highest_exam)
```

# 4. Which era had more paper leak cases, and how did their impact differ?

```python
# UPA and NDA Comparison

era_analysis = df.groupby('era').agg(
    leak_cases=('incident_id', 'count'),
    affected_aspirants=('aspirants_affected', 'sum'),
    arrests=('arrests', 'sum'),
    convictions=('convictions', 'sum')
).reset_index()

display(era_analysis)
```

# 5. Which year had the highest number of affected aspirants?

```python
# Year with Highest Affected Aspirants

aspirant_analysis = df.groupby('year')['aspirants_affected'].sum().reset_index()

highest_aspirant_year = aspirant_analysis.loc[
    aspirant_analysis['aspirants_affected'].idxmax()
]

display(highest_aspirant_year)
```

# 6. Is there a relationship between arrests and convictions?

```python
# Relationship Between Arrests and Convictions

arrest_conviction = df[['arrests', 'convictions']].copy()

correlation = arrest_conviction['arrests'].corr(
    arrest_conviction['convictions']
)

print("Correlation between arrests and convictions:", correlation)
```

# 7. How did paper leak cases change over the years?

```python
# Paper Leak Trend Over the Years

yearly_trend = df.groupby('year').size().reset_index(
    name='leak_count'
)

display(yearly_trend)
```
